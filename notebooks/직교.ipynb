```python
import numpy as np

# 평균 중심화(centering)한 두 변수의 상관계수 = 두 벡터의 코사인 각도
# 상관 0 = 코사인 0 = 90도 = 직교(orthogonal)

np.random.seed(42)
x = np.random.randn(100)
y = np.random.randn(100)  # x와 무관하게 생성

xc = x - x.mean()
yc = y - y.mean()

corr = np.corrcoef(x, y)[0, 1]                                    # 통계적 정의
cos_angle = np.dot(xc, yc) / (np.linalg.norm(xc) * np.linalg.norm(yc))  # 기하학적 정의

print(f"상관계수      : {corr:.4f}")
print(f"코사인(벡터각): {cos_angle:.4f}")
print(f"같은 값인가   : {np.isclose(corr, cos_angle)}")
```

```python
# 직교(상관 0)인데 명백히 종속인 반례
x = np.linspace(-3, 3, 200)
y = x**2                          # 완벽한 함수 관계
print(f"상관계수: {np.corrcoef(x, y)[0, 1]:.4f}")  # ≈ 0
# 선형 상관은 0이지만 y는 x로 완전히 결정된다 → 직교가 독립을 뜻하지 않는다
```
